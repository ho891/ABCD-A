```python
import math

# 주어진 값
theta_deg = 45  # 기울기 각도 (도)
r = 20  # 예시 커브 반지름 (단위: m, 필요 시 변경 가능)
g = 9.8  # 중력 가속도 (m/s^2)

# (1) tan θ = v^2 / (r * g) => v^2 = r * g * tan θ
theta_rad = math.radians(theta_deg)
tan_theta = math.tan(theta_rad)

v_squared = r * g * tan_theta
v = math.sqrt(v_squared)

print(f"(1) θ = {theta_deg}°일 때, 속력 v = {v:.2f} m/s")

# (2) 구심가속도 a_c = v^2 / r
a_c = v_squared / r
print(f"(2) 구심가속도 a_c = {a_c:.2f} m/s²")
```
