```python
i = 0
result1 = 0
while i < 100:
    i = i + 1
    if i % 2 == 0:
        result1 = result1 + i

print('1번 방법 : {0}'.format(result1))
```

```python
# 2번 방법 (break 이용)
j = 0
result2 = 0
while True:
    if j > 100:
        break

    j = j + 1
    if j % 2 == 0:
        result2 = result2 + j

print('2번 방법 (break) : {0}'.format(result2))
```

```python
# 3번 방법 (continue 이용)
k = 0
result3 = 0
while k < 100:
    k = k + 1
    if k % 2 != 0:  # 홀수라면 continue로 건너뜁니다
        continue

    result3 = result3 + k  # 짝수를 더합니다

print('3번 방법 (continue) : {0}'.format(result3))  # 결과 출력
```
