```python
import math

# 주어진 값
mass = 0.15  # kg (질량, 실제 계산엔 사용되지 않음)
radius = 0.60  # m (반지름)
frequency = 2.0  # Hz (1초에 2회전)

# (a) 주기 T 계산
T = 1 / frequency
print(f"(a) 주기 T = {T:.2f} s")

# (b) 속력 v 계산: v = 2 * pi * r / T
v = 2 * math.pi * radius / T
print(f"(b) 속력 v = {v:.2f} m/s")

# (c) 구심가속도 a_c 계산: a_c = v^2 / r
a_c = v ** 2 / radius
print(f"(c) 구심가속도 a_c = {a_c:.2f} m/s²")
```
