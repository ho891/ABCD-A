```python
# 주어진 값
r = 500          # 반지름 (m)
t = 10.5         # 가속 시간 (s)
v_final = 32     # 최종 속력 (m/s)
v_current = 35   # (b) 상황 속력 (m/s)

# (a) 접선 가속도 계산
a_t = v_final / t
print(f"(a) 접선 가속도 a_t = {a_t:.2f} m/s²")

# (b) 지름 방향(구심) 가속도 계산
a_r = v_current ** 2 / r
print(f"(b) 지름 가속도 a_r = {a_r:.2f} m/s²")
```
