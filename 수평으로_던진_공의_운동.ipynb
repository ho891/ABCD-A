```python
import math
```

수평으로 던진 공의 운동

```python
# 초기속도 입력
vo=float(input())
print("초기속도:",vo,"m/s")
```

```python
#초기 각도 입력
angle_degree=float(input())
print("초기각도:",angle_degree,"도")

#각도를 radian으로바꾸기
angle_degree=math.radians(angle_degree)
print("초기각도:",angle_degree,"radian")
```

```python
#X 성분의 속도를 구하시오
vOx = math.cos(angle_degree) * vo
print("X축의 초기속도:", vOx, "m/s")
```

```python
#Y 성분의 속도를 구하시오
vOy = math.sin(angle_degree) * vo
print("Y축의 초기속도:", vOy, "m/s")
```

```python
#최고점 높이
g= 9.8
h_max = (vOy**2) / (2 * g)
print("최고점 높이:", h_max, "m")
```

```python
#최고점 도달시간
t_max = vOy / g
print("최고점 도달 시간:", t_max, "초")
```

```python
# 수평으로 이동한 거리 구하기
t_total = 2 * t_max  # 전체 비행 시간
dist = vOx * t_total
print("수평 이동 거리:", dist, "m")
```

```python
# 지면에 닿을 때의 Vx, Vy와 V 구하기
vy_end = -vOy
v_end = math.sqrt(vOx**2 + vy_end**2)
print("지면에 닿을 때 x 방향 속도:", vOx, "m/s")
print("지면에 닿을 때 y 방향 속도:", vy_end, "m/s")
print("지면에 닿을 때 속도 크기:", v_end, "m/s")
print("속도 벡터 성분: (", vOx, ",", vy_end, ")")
```
