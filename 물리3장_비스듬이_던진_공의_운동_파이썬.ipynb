```python
import math

# 초기 조건
v0 = 100  # 초기 속도 (m/s)
angle_deg = 37  # 발사 각도 (deg)
angle_rad = math.radians(angle_deg)  # 각도를 라디안으로 변환
y0 = 140  # 초기 높이 (m)
g = 9.80  # 중력 가속도 (m/s^2)

# 수평 및 수직 성분 분해
v0x = v0 * math.cos(angle_rad)
v0y = v0 * math.sin(angle_rad)

print(f"(a) 수직 낙하 시간 구하기:")
# y = y0 + v0y*t - 0.5*g*t^2 = 0 되는 시간 구하기 (2차방정식)
# 0 = y0 + v0y*t - 0.5*g*t^2
a = -0.5 * g
b = v0y
c = y0

# 근의 공식 사용
discriminant = b**2 - 4*a*c
if discriminant < 0:
    print("해가 없습니다.")
else:
    t1 = (-b + math.sqrt(discriminant)) / (2 * a)
    t2 = (-b - math.sqrt(discriminant)) / (2 * a)
    t = max(t1, t2)  # 양의 시간만 고려
    print(f"총 시간 t = {t:.2f} s")

    print("\n(b) 수평 거리 x 구하기:")
    x = v0x * t
    print(f"수평 거리 x = {x:.2f} m")

    print("\n(c) 점 P에서의 속도 성분 구하기:")
    vx = v0x  # 수평 속도는 변하지 않음
    vy = v0y - g * t  # 최종 수직 속도
    print(f"vx = {vx:.2f} m/s, vy = {vy:.2f} m/s")

    print("\n(d) 속도의 크기 구하기:")
    v = math.sqrt(vx**2 + vy**2)
    print(f"속력 v = {v:.2f} m/s")

    print("\n(e) 속도 벡터가 이루는 각도 구하기:")
    theta_rad = math.atan2(vy, vx)
    theta_deg = math.degrees(theta_rad)
    print(f"각도 θ = {theta_deg:.2f}°")
```
